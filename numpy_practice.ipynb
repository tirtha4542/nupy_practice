Task 1: Modify the code to create a 4x4 matrix of random integers between 10 and 50, then set all
values greater than 40 to 0.

```python
import numpy as np

array = np.random.randint(10, 50, size=(4, 4))
print("Original array:\n",array)
array[array>40]=0
print("Modified array:\n",array)
```

Task 5:Task: Modify the code to create a boxplot of total_bill grouped by day, with points colored by
sex.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", data=tips,hue="sex")
plt.title("Box Plot of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.legend(title="sex")
plt.show()


```

Task 6:Task: Modify the code to create a heatmap of the correlation matrix for numerical columns in the
tips dataset, with annotations.

```python
select = tips.select_dtypes(include=['number'])
correlation = select.corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

task 7:Task: Modify the code to create a scatter plot of total_bill vs. tip, with points colored by day and
sized by size.


```python
import plotly.express as px

tips = px.data.tips()

fig = px.scatter(tips, x="total_bill", y="tip", color="day", title="Scatter Plot of Total Bill vs Tip",size="size",hover_data=["time","smoker"])
fig.show()

```

Task 8:Modify the code to create a bar plot showing the average total_bill per day, with bars
colored by sex.


```python
gro = tips.groupby(["day","sex"])["total_bill"].mean().reset_index()
gro
fig2 = px.bar(gro,x="day",y="total_bill",title="Average Total Bill by Day",barmode="group",color="sex")
fig2.show()
```
